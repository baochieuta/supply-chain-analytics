```python
#drop redundant column
cols_to_drop = ['Customer Email', 'Customer Fname', 'Customer Lname', 'Customer Password', 'Customer Street', 'Product Description', 'Order Zipcode', 'Category Id', 'Product Category Id', 'Order Item Cardprod Id', 'Department Id', 'Product Image']

df = df.drop(columns = cols_to_drop)
df.shape
```

```python
#change date type
df['order date (DateOrders)'] = pd.to_datetime(df['order date (DateOrders)'])
df['shipping date (DateOrders)'] = pd.to_datetime(df['shipping date (DateOrders)'])
                                                                                           
```

```python
df.dtypes[['order date (DateOrders)', 'shipping date (DateOrders)']]    
```

```python
#verify the shipping date
df['calculated_shipping_days'] = (df['shipping date (DateOrders)'] - df['order date (DateOrders)']).dt.days
df[['order date (DateOrders)', 'shipping date (DateOrders)', 'calculated_shipping_days', 'Days for shipping (real)']].head()
```

```python
#check yes or no if calculated shipping days is same as days for shipping
(df['calculated_shipping_days'] == df['Days for shipping (real)']).all()
```

```python
mismatches = df['calculated_shipping_days'] != df['Days for shipping (real)']
mismatches.sum()

#after two check, we see that calculated_shippingdays is not good
```

```python
mismatches = df['calculated_shipping_days'] != df['Days for shipping (real)']
df[mismatches][['order date (DateOrders)', 'shipping date (DateOrders)', 'calculated_shipping_days', 'Days for shipping (real)']].head(10)
                                                                                                                        
```

```python
#confirm the difference is by definition, not just error
diff = df.loc[mismatches, 'Days for shipping (real)'] - df.loc[mismatches, 'calculated_shipping_days']
diff.value_counts()
```

Verified Days for shipping (real) against a manually calculated shipping duration. 2.6% of rows (4,657/180,519) sho a discrepancy of exactly 1 day, consistently in the same direction - caused by the original column counting calendar days crossed rather than elapsed 24-hour periods(e.g., an order place 11:30 PM and shipped 11:30 AM the next day spans 2 calendar dates but only 12 actual hours). both are valud definitions; we retain the original Days for shipping (real) column since it's what the Late_delivery_risk label was presumably built against, and drop our calculated column

```python
#drop calculated_shipping_days
df = df.drop(columns = ['calculated_shipping_days'])
df.shape
```

```python
df['Order City'].nunique()
```

Order City retained despite high cardinality (3,597 unique values) - useful for dashboard drill-downs (top cities by order volume/ late-delivery rate/ product category), but will be excluded from delivery-risk model features due to cardinality

```python
df[df['Customer Zipcode'].isna()]
```

```python
#drop the row with NaN customer zipcode
df = df.dropna(subset = ['Customer Zipcode'])
df.shape
```

Customer Zipcode has three entries that are NaN. Those 3 entries might have been a misinterpretation for other the LATAM market (as they write EE.UU instead of United States). The entries made up of only 0.0017% of the data, instead of trying to guess correctly and reconstruct the zipcode, we decided to get rid of the entries
# Better version
Customer Zipcode is missing for 3 rows (0.0017% of the data). Inspecting these rows also revealed apparent column misalignment in the geography fields (e.g., Customer State holding what looks like a zip code, Customer Country showing "EE.UU." — Spanish for "United States," suggesting a LATAM-market entry issue). Given the negligible sample size and that correcting this would require guessing rather than recovering real values, these 3 rows were dropped rather than imputed.
